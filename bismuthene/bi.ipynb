```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.pyplot import figure

band_data_up = np.loadtxt('bands.dat.gnu')
k0 = np.unique(band_data_up[:,0]) 
bands_up = np.reshape(band_data_up[:,1],(-1,len(k0)))

band_data0 = np.loadtxt('bands.dat.gnu')
k0 = np.unique(band_data0[:,0]) 
bands0 = np.reshape(band_data0[:,1],(-1,len(k0)))

fermi_energy = -1.7897

figure(figsize=(10, 8))

for band in range(len(bands0)):
    plt.plot(k0, bands_up[band,:]-fermi_energy, lw=0.75, color='blue')
    #plt.plot(k0,bands0[band,:]-fermi_energy, lw=0.75, color='red')
plt.xlim(min(k0),max(k0))

plt.axvline(x=0,lw=0.5,color='black')
plt.axvline(x=0.5000,lw=0.5,color='black')
plt.axvline(x=1.2806,lw=0.5,color='black')
plt.axvline(x=1.7806,lw=0.5,color='black')
plt.axvline(x=2.5613,lw=0.5,color='black')

plt.xticks(ticks=[0,0.5000,1.2806,1.7806,2.5613],
           labels=['5-0','$\Gamma$','0-5','5-5','5-0'])

plt.axhline(y=0,ls='--',c='gray',lw=0.5)
plt.ylim(-6,6)
plt.ylabel(r'$E - E_f$')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

data_s = np.loadtxt('atom_Bi_s.dat')
data_p = np.loadtxt('atom_Bi_p.dat')
data_d = np.loadtxt('atom_Bi_d.dat')

energy , dos_s = data_s.T
_, dos_p = data_p.T
_, dos_d = data_d.T

fermi_energy = -1.7897

plt.figure(figsize=(8,6))
plt.plot(energy, dos_s, label='s', color='blue')
plt.plot(energy, dos_p, label='p', color='orange')
plt.plot(energy, dos_d, label='d', color='green')
plt.axvline(x=fermi_energy, ls='--', color='red', label='Fermi level')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.pyplot import figure

band_data = np.loadtxt('supercell-bands.dat.gnu')
k0 = np.unique(band_data[:,0]) 
bands = np.reshape(band_data[:,1],(-1,len(k0)))

fermi_energy = -2.9619

figure(figsize=(10, 8))

for band in range(len(bands)):
    plt.plot(k0, bands[band,:]-fermi_energy, lw=0.75, color='blue')

plt.xlim(min(k0),max(k0))

plt.axvline(x=0,lw=0.5,color='black')
plt.axvline(x=5.0234,lw=0.5,color='black')
plt.axvline(x=10.0469,lw=0.5,color='black')
plt.axvline(x=15.0703,lw=0.5,color='black')

plt.xticks(ticks=[0,5.0234,10.0469,15.0703],
           labels=['M',r'$\Gamma$','M',r'$\Gamma$'])

plt.axhline(y=0,ls='--',c='gray',lw=0.5)
plt.ylim(-6,6)
plt.ylabel(r'$E - E_f$')
plt.show()
```
