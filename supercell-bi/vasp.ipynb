```python
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt('REFORMATTED_BAND.dat')

k = data[:, 0]
bands = data[:, 1:]
num_bands = bands.shape[1]
print(num_bands)
high_symmetry_points = [0,0.718,1.435]  
high_symmetry_labels = [ 'X', r'$\Gamma$', 'X']

plt.figure(figsize=(10, 8))
for i in range(num_bands):
    plt.plot(k, bands[:, i], color='blue',lw=0.75)

plt.plot(k,bands[:,392], color='red',lw=1.5)

plt.axhline(0, color='black', linestyle='--', lw=0.5)
for i in range(len(high_symmetry_points)):
    plt.axvline(high_symmetry_points[i], color='black', linestyle='--', lw=0.5)

plt.xticks(high_symmetry_points, high_symmetry_labels, fontsize=16)
plt.ylabel('Energy (eV)', fontsize=16)
plt.title("SOC ON")
plt.xlim(min(k), max(k))
plt.ylim(-2, 2)
```

```python

import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Load data
data = np.loadtxt("SPINTEXTURE_2D_GAMMA_CENTER.dat", comments="#")
kx, ky, energy, sx, sy, sz = data.T

# Create 3D scatter plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
p = ax.scatter(kx, ky, energy, c=sz, cmap='coolwarm', marker='o')
plt.quiver(kx, ky, energy, sx, sy, 0, length=0.03, normalize=True, color='black', alpha=0.5)
ax.set_xlabel("kₓ")
ax.set_ylabel("kᵧ")
ax.set_zlabel("Energy (eV)")
ax.set_title("3D Band Structure with Spin-z Coloring")
fig.colorbar(p, ax=ax, label="S_z")
ax.view_init(elev=0, azim=0)
plt.show()
```
