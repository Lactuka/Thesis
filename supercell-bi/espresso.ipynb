```python
import numpy as np
import matplotlib.pyplot as plt

band_data = np.loadtxt('wosoc-bands.dat.gnu')
k_points = np.unique(band_data[:, 0])
num_bands = int(band_data.shape[0] / len(k_points))
energies = band_data[:, 1].reshape((num_bands, len(k_points)))
fermi_level = -0.2925
high_symmetry_points = [0.0000,0.5000,1.3660,1.8660]
high_symmetry_labels = ['Γ', 'M', 'X', 'K']

wannier_data = np.loadtxt('wosoc-wannier_band.dat')
k_points_wannier = np.unique(wannier_data[:, 0])
num_bands_wannier = int(wannier_data.shape[0] / len(k_points_wannier))
energies_wannier = wannier_data[:, 1].reshape((num_bands_wannier, len(k_points_wannier)))

k_points_wannier_reduced = k_points_wannier[0:373]
energies_wannier_reduced = energies_wannier[:, 0:373]

#adjust k scale
k_points_wannier_reduced = k_points_wannier_reduced * (max(k_points)/max(k_points_wannier_reduced))


plt.figure(figsize=(10, 8))

for i in range(num_bands):
    plt.plot(k_points, energies[i]-fermi_level, color='blue', linewidth=0.75 , label ='DFT' if i==0 else "")

for i in range(num_bands_wannier):
    plt.plot(k_points_wannier_reduced, energies_wannier_reduced[i]-fermi_level, color='red', linewidth=0.75 , linestyle='--',label ='Wannier' if i==0 else "")


for i in range(len(high_symmetry_points)):
    plt.axvline(x=high_symmetry_points[i], color='black', linestyle='--',linewidth=0.5)

print(num_bands)
plt.axhline(y=0, color='red', linestyle='--', linewidth=0.5)
plt.xticks(high_symmetry_points, high_symmetry_labels)
plt.ylabel('Energy (eV)')
plt.title('Electronic Band Structure')
plt.xlim(min(high_symmetry_points), max(high_symmetry_points))
plt.ylim(-20, 10)

```
