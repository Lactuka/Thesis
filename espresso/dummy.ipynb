```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.pyplot import figure

band_data_up = np.loadtxt('bands-up.dat.gnu')
k0 = np.unique(band_data_up[:,0]) 
bands_up = np.reshape(band_data_up[:,1],(-1,len(k0)))

band_data0 = np.loadtxt('bands.dat.gnu')
k0 = np.unique(band_data0[:,0]) 
bands0 = np.reshape(band_data0[:,1],(-1,len(k0)))

fermi_energy = -0.9145

figure(figsize=(10, 8))

for band in range(len(bands0)):
    plt.plot(k0, bands_up[band,:]-fermi_energy, lw=0.75, color='blue')
    plt.plot(k0,bands0[band,:]-fermi_energy, lw=0.75, color='red')
plt.xlim(min(k0),max(k0))

plt.axvline(x=0,lw=0.5,color='black')
plt.axvline(x=0.6717,lw=0.5,color='black')
plt.axvline(x=1.2534,lw=0.5,color='black')
plt.axvline(x=1.5892,lw=0.5,color='black')
plt.axvline(x=2.2609,lw=0.5,color='black')

plt.xticks(ticks=[0,0.6717,1.2534,1.5892,2.2609],
           labels=['K','$\Gamma$','M','K','$\Gamma$'])

plt.axhline(y=0,ls='--',c='gray',lw=0.5)
plt.ylabel(r'$E - E_f$')
plt.ylim(-3,3)
plt.title('Band structure of buckled BiSb monolayer')
plt.show()
```
