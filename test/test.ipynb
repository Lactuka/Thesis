```python
import numpy as np
import matplotlib.pyplot as plt

# Load data while skipping comment lines starting with "#"
data = np.loadtxt('SPINTEXTURE_2D_GAMMA_CENTER.dat', comments='#')

# Extract columns
kx = data[:, 0]
ky = data[:, 1]
energy = data[:, 2]
sx = data[:, 3]
sy = data[:, 4]
sz = data[:, 5]

# Infer grid size (e.g., 30x30)
grid_size = int(np.sqrt(len(kx)))  # or manually set to 30 if it's not square

# Reshape data for grid plotting
kx_grid = kx.reshape((grid_size, grid_size))
ky_grid = ky.reshape((grid_size, grid_size))
sx_grid = sx.reshape((grid_size, grid_size))
sy_grid = sy.reshape((grid_size, grid_size))
sz_grid = sz.reshape((grid_size, grid_size))

# Normalize arrows for better visual clarity (optional)
norm = np.sqrt(sx_grid**2 + sy_grid**2)
norm[norm == 0] = 1  # avoid divide-by-zero
sx_norm = sx_grid / norm
sy_norm = sy_grid / norm

# Plot spin texture (Sx, Sy) as arrows
plt.figure(figsize=(8, 6))
plt.quiver(kx_grid, ky_grid, sx_norm, sy_norm, sz_grid, cmap='coolwarm', scale=30)
plt.xlabel('kₓ')
plt.ylabel('kᵧ')
plt.title('Spin Texture (Sₓ, Sᵧ) colored by S_z')
plt.colorbar(label='S_z')
plt.xlim(min(kx), max(kx))
plt.ylim(min(ky), max(ky))
plt.axis('equal')
plt.tight_layout()
plt.show()
```

```python
def read_hamiltonian(path):

    """
    Read hopping matrix element from the wannier90 output file:wannier90_hr.dat
    wan_num: number of wannier functions
    wsc_num: number of wigner-sitez cells
    wsc_count: variables related to degeneracy
    """

    with open(path,"r") as f:
        lines=f.readlines()

    wan_num=int(lines[1]); wsc_num=int(lines[2])
    ski_row_num=int(np.ceil(wsc_num/15.0))      # skip row numbers
    wsc_count=[]
    for i in range(ski_row_num):
        wsc_count.extend(list(map(int,lines[i+3].split())))

    wsc_tot=np.zeros((wan_num**2*wsc_num,3))
    tem_tot=np.zeros((wan_num**2*wsc_num,2))
    for i in range(wan_num**2*wsc_num):
        wsc_tot[i,:]=list(map(int,lines[3+ski_row_num+i].split()[:3]))
        tem_tot[i,:]=list(map(float,lines[3+ski_row_num+i].split()[5:]))

    wsc_idx=wsc_tot[0:-1:wan_num**2,:]   # the translational vector between wigner-sitez cells
    hop_mat=np.reshape(tem_tot[:,0]+1j*tem_tot[:,1],[wan_num,wan_num,wsc_num],order='F')

    return hop_mat,wsc_idx,wsc_count,wan_num

# Example usage
filename = "wannier_hr.dat"
hop_mat,wsc_idx,wsc_count,wan_num = read_hamiltonian(filename)


def H_of_k(kvec, hop_mat, wsc_idx, wsc_count):
    """
    Construct H(k) from real-space hoppings.
    kvec: np.array([kx,ky,kz])
    hop_mat: (wan_num, wan_num, wsc_num)
    wsc_idx: (wsc_num, 3)
    wsc_count: list of degeneracies
    """
    wan_num = hop_mat.shape[0]
    Hk = np.zeros((wan_num, wan_num), dtype=complex)
    for R, count, H_R in zip(wsc_idx, wsc_count, hop_mat.transpose(2,0,1)):
        phase = np.exp(1j * np.dot(kvec, R))
        Hk += count * H_R * phase
    return Hk / sum(wsc_count)

```

```python
import numpy as np
import matplotlib.pyplot as plt

# === Load the TB Hamiltonian from wannier_hr.dat ===
filename = "wannier_hr.dat"
hop_mat, wsc_idx, wsc_count, wan_num = read_hamiltonian(filename)

# === Define k-path ===
# Example for a square lattice Γ -> X -> M -> Γ
# Replace with the correct high-symmetry points of your system
k_points = {
    "G": [0.0, 0.0, 0.0],
    "K": [0.333, 0.3333, 0.0],
    "M": [0.5, 0.5, 0.0],
}
path = ["K", "G", "M", "K"]

# Interpolation (number of k-points between each segment)
nseg = 50

k_path = []
k_labels = []
distances = [0.0]

for i in range(len(path)-1):
    k_start = np.array(k_points[path[i]])
    k_end   = np.array(k_points[path[i+1]])
    for j in range(nseg):
        frac = j / nseg
        k_vec = (1-frac)*k_start + frac*k_end
        k_path.append(k_vec)
        if len(k_path) > 1:
            dk = np.linalg.norm(np.array(k_path[-1]) - np.array(k_path[-2]))
            distances.append(distances[-1] + dk)
    k_labels.append((distances[-1], path[i]))
k_labels.append((distances[-1], path[-1]))
k_path = np.array(k_path)

# === Build Hamiltonian H(k) and get eigenvalues ===
bands = []

for k in k_path:
    Hk = np.zeros((wan_num, wan_num), dtype=complex)
    for iR, R in enumerate(wsc_idx):
        phase = np.exp(1j * 2*np.pi * np.dot(k, R))
        Hk += hop_mat[:, :, iR] * phase * wsc_count[iR]
    eigvals = np.linalg.eigvalsh(Hk)
    bands.append(np.real(eigvals))

bands = np.array(bands)

# === Plot ===
plt.figure(figsize=(6,6))
for n in range(wan_num):
    plt.plot(distances, bands[:, n], 'k-')

# High-symmetry lines
for x, label in k_labels:
    plt.axvline(x, color='gray', linestyle='--')
    plt.text(x, plt.ylim()[0]-0.5, label, ha='center')

plt.ylabel("Energy (eV)")
plt.xlabel("k-path")
plt.title("Band Structure from Wannier Hamiltonian")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ============================
# Step 1: Read wannier90_hr.dat
# ============================
filename = "wannier_hr.dat"

with open(filename, 'r') as f:
    lines = f.readlines()

# Number of Wannier functions (orbitals)
num_wann = int(lines[1])
# Number of R-vectors
num_R = int(lines[2])

# Read degeneracies
degeneracies = []
for i in range(int(np.ceil(num_R / 15))):
    degeneracies.extend([int(x) for x in lines[3+i].split()])
degeneracies = np.array(degeneracies)

# Extract hopping data
hopping_data = []
start = 3 + int(np.ceil(num_R / 15))
for line in lines[start:]:
    parts = line.split()
    if len(parts) == 0:
        continue
    R = [int(parts[0]), int(parts[1]), int(parts[2])]
    i = int(parts[3]) - 1   # orbital index (0-based)
    j = int(parts[4]) - 1
    t_real = float(parts[5])
    t_imag = float(parts[6])
    hopping_data.append((R, i, j, t_real + 1j*t_imag))

# ============================
# Step 2: Construct H(k)
# ============================
def Hk(kvec):
    """Build the Hamiltonian for a given k-vector"""
    H = np.zeros((num_wann, num_wann), dtype=complex)
    for n, (R, i, j, t) in enumerate(hopping_data):
        phase = np.exp(1j * 2*np.pi * np.dot(kvec, R)) / degeneracies[n]
        H[i, j] += t * phase
    return H

# ============================
# Step 3: Define k-path
# ============================
# Example for a square lattice: Γ -> X -> M -> Γ
k_points = {
    "Γ": [0.0, 0.0, 0.0],
    "X": [0.5, 0.0, 0.0],
    "M": [0.5, 0.5, 0.0]
}

path = ["Γ", "X", "M", "Γ"]
nk = 100  # number of k-points per segment

kvecs = []
k_dist = [0]  # cumulative distance
k_labels = [0]
label_positions = [0]

for i in range(len(path)-1):
    start = np.array(k_points[path[i]])
    end   = np.array(k_points[path[i+1]])
    for j in range(nk):
        frac = j / nk
        k = start + frac*(end - start)
        kvecs.append(k)
        if len(kvecs) > 1:
            dk = np.linalg.norm(np.array(kvecs[-1]) - np.array(kvecs[-2]))
            k_dist.append(k_dist[-1] + dk)
    label_positions.append(k_dist[-1])

kvecs = np.array(kvecs)
k_dist = np.array(k_dist)

# ============================
# Step 4: Diagonalize H(k)
# ============================
bands = []
for k in kvecs:
    eigvals = np.linalg.eigvalsh(Hk(k))
    bands.append(eigvals.real)

bands = np.array(bands)

# ============================
# Step 5: Plot bands
# ============================
plt.figure(figsize=(6,5))
for n in range(num_wann):
    plt.plot(k_dist, bands[:,n], 'b-')

plt.xticks(label_positions, path)
plt.ylabel("Energy (eV)")
plt.title("Wannier Tight-Binding Band Structure")
plt.axhline(0, color='k', linewidth=0.5, linestyle="--")
plt.show()

```

```python
import numpy as np
import matplotlib.pyplot as plt

elemented_data = np.loadtxt("phs-PBAND_ELEMENTS_SOC.dat")

# Define elements and their corresponding columns (starting from column 2)
# Adjust this based on your actual data structure
element_names = ['P', 'Bi', 'Sb']  # Add more as needed
element_columns = [2, 3, 4]  # Column indices for each element

# Define colors for each element
colors = ['red', 'blue', 'green']

# Find which element is dominant at each k-point
dominant_elements = np.argmax(elemented_data[:, element_columns], axis=1)

# Find where the dominant element changes
color_changes = np.diff(dominant_elements)
change_indices = np.where(color_changes != 0)[0] + 1

# Add start and end indices
segment_indices = np.concatenate(([0], change_indices, [len(elemented_data)]))

# Keep track of which elements have been labeled
labeled_elements = set()

# Plot each segment
for i in range(len(segment_indices) - 1):
    start = segment_indices[i]
    end = segment_indices[i + 1]
    
    dominant_idx = dominant_elements[start]
    color = colors[dominant_idx]
    label = element_names[dominant_idx]
    
    # Only add label for the first occurrence of each element
    if label not in labeled_elements:
        plt.plot(elemented_data[start:end+1, 0], elemented_data[start:end+1, 1], 
                color=color, label=label, linewidth=0.5)
        labeled_elements.add(label)
    else:
        plt.plot(elemented_data[start:end+1, 0], elemented_data[start:end+1, 1], 
                color=color, linewidth=0.5)

# High symmetry points
high_sym_points = [0, 0.725,1.563,2.400, max(elemented_data[:,0])]
high_sym_labels = ['X', 'Γ', 'M', 'Γ', 'Y']

plt.xticks(high_sym_points, high_sym_labels)
plt.xlim(min(elemented_data[:,0]), max(elemented_data[:,0]))
plt.legend()
plt.title("Element-Projected Band Structure")
plt.ylabel('Energy')
plt.ylim(-2, 2)
plt.show()
```
